```python
                          # N_RE = (pvD)/u                                    (1)

                   # VARIABLE EXPLANATION WITH UNITS
# p = density in kg/m^3, where p_O2, p_Heliox, p_Air are the densities of ...
# ... oxygen, heliox 80:20 and air, respectively.
# v = Velocity in m/s.
# D = Trachea internal diameter in m [1].
# u = Dynamic viscosity in kg/m*s, where u_O2, u_Heliox and u_Air are the ...
# ... dynamic viscosities of oxygen, heliox 80:20 and air, respectively.

p_O2 = 1.241 
p_Heliox = 0.378 # [3].
p_Air = 1.138 # [3].

Q = (10/60)*0.001 # Volumetric flow rate converted from 10 L/min into m^3/s [2].
D = 19.5/1000 # Trachea internal diameter utilizing standardized adult ...
# ... Trachea model converted from 19.5 mm into m [1].
pi = 3.14 # Dimensionless pi.
A = (pi/4)*D**2 # Cross-sectional area in m^2.
v = Q/A

u_O2 = 2.12E-5
u_Heliox = 2.036E-5 # At 37 degrees C [3].
u_Air = 1.885E-5 # At 37 degrees C [3].

N_Re_O2 = (p_O2*v*D)/(u_O2)
N_Re_Heliox = (p_Heliox*v*D)/(u_Heliox)
N_Re_Air = (p_Air*v*D)/(u_Air)
print(f"N_Re of oxygen (O2) = {N_Re_O2:.3}")
print(f"N_Re of Heliox (80/20) = {N_Re_Heliox:.3}")
print(f"N_Re of Air = {N_Re_Air:.3}")
```
